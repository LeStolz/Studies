```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np
```

```python
T = 1
K = 10
n = 10
m = 1000

def simulated_function(t):
    return np.sin(6 * np.pi * t) + np.random.normal(0, 0.25, size=t.shape)

t = np.linspace(0, T, m)
y = np.array(simulated_function(t))

fig, ax = plt.subplots(figsize=(18, 6))

ax.plot(t, y)

plt.show()
```

```python
def phi(k, t):
    return t ** k

X = np.array([phi(k, t) for k in range(K)]).T

model = LinearRegression()
model.fit(X, y)

predictions = model.predict(X)

fig, ax = plt.subplots(figsize=(18, 6))

ax.plot(t, y)
ax.plot(t, predictions)

plt.show()
```
